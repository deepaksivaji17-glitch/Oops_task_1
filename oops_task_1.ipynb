# 1.Abstract

```python
from abc import ABC,abstractmethod
```

```python
from abc import ABC, abstractmethod
class Employee(ABC):
    def __init__(self,name):
        self.name=name   
    def login(self):
        pass    
    def view_tasks(self):
        print(f"{self.name} is viewing the task")
    @abstractmethod
    def work(self):
        pass
class TeamMember(Employee):
    def submit_tasks(self):
        print(f"{self.name} task has submitted")
    def work(self):
        pass    
class Manager(Employee):
    def assign_tasks(self):
        print(f"{self.name} has assigned the task")
    def work(self):
        pass    
deepak = TeamMember("deepak")
dheekshana = Manager("dheekshana")

deepak.login()
deepak.submit_tasks()
deepak.view_tasks()
deepak.name
deepak.work()

dheekshana.login()
dheekshana.assign_tasks()
dheekshana.view_tasks()
dheekshana.name
dheekshana.work()
```

# 2.inheritance

```python
class Employee:
    def __init__(self,name):
        self.name=name
    def login(self):
        print(f"{self.name} has logged in")
    def view_tasks(self):
        print(f"{self.name} is viewing the task")
    def work(self):
        pass       
class TeamMember(Employee):
    def submit_tasks(self):
        print(f"{self.name} task has submitted")
    def work(self):
        pass       
class Manager(Employee):
    def assign_tasks(self):
        print(f"{self.name} has assigned the task")
    def work(self):
        pass       
deepak = TeamMember("deepak")
dheekshana = Manager("dheekshana")

deepak = TeamMember("deepak")
dheekshana = Manager("dheekshana")

deepak.login()
dheekshana.login()
deepak.work()
dheekshana.work()
deepak.submit_tasks()
dheekshana.assign_tasks()
deepak.submit_tasks()
dheekshana.view_tasks()
```

# 3.Polymorphism

```python
class Employee:
    def __init__(self,name):
        self.name=name
    def login(self):
        print(f"{self.name} has logged in")
    def view_tasks(self):
        print(f"{self.name} is viewing the task")
class TeamMember(Employee):
    def submit_tasks(self):
        print(f"{self.name} task has submitted")
    def work(self):
        print(f"{self.name} is completing task")      
class Manager(Employee):
    def assign_tasks(self):
        print(f"{self.name} has assigned the task")
    def work(self):
        print(f"{self.name} is managing the task")       
        
deepak = TeamMember("deepak")
dheekshana = Manager("dheekshana")

deepak.work()
dheekshana.work()
```
